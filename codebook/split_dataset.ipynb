```python
import os 
import numpy as np 
import pandas as pd 
import random

%autosave 300
```

```python
# Set up our directories 
currPath = os.getcwd()
dataPath = os.path.abspath(os.path.join(currPath, os.pardir, 'data/playlist_subset'))

# From the last dataframe we have, we will choose 1k playlists that we will consider apart of our training set
playlist_files = [x for x in os.listdir(dataPath) if x.startswith('playlist_sub')]
playlist_files.sort()

# Import the last dataframe 
play_data = pd.read_csv(os.path.join(dataPath, playlist_files[-1]), compression='gzip')

# First, get the unique playlist values, and then shuffle that list
all_pid = np.unique(play_data['pid'].values)
all_pid = random.sample(list(all_pid), len(all_pid))

train_subset = []
for a in all_pid[:1000]: 
    curr_playlist = play_data[play_data['pid'] == a]
    test_subset.append(curr_playlist)

    # Then, remove this from our playlist data overall
    play_data = play_data[play_data['pid'] != a]

# Initialize our testing dataframe
train_subset = pd.concat(train_subset)

# Then, save both dataframes accordingly 
train_fn = playlist_files[-1].replace('_4.csv', '_4_train.csv')
test_fn = playlist_files[-1].replace('_4.csv', '_4_test.csv')

train_subset.to_csv(os.path.join(dataPath, train_fn), index=False, compression = 'gzip') # training data
play_data.to_csv(os.path.join(dataPath, test_fn), index=False, compression = 'gzip') # test data
```
