```python
from gensim.models import Word2Vec
from scipy.spatial.distance import cosine
from tqdm import tqdm
import numpy as np
import pandas as pd
import random

emotion_cols = ["admiration","amusement","anger","annoyance","approval","caring","confusion", "curiosity", "desire", "disappointment", "disapproval","disgust", "embarrassment", "excitement", "fear","gratitude","grief","joy", "love","nervousness","optimism","pride","realization","relief","remorse","sadness","surprise","neutral"]

# Word2Vec training
playlist_sequences = df.groupby("pid")["track_id"].apply(list).tolist()
model = Word2Vec(sentences=playlist_sequences,
                 vector_size=100, window=8, min_count=5,
                 workers=4, sg=1)

# Precompute lookups
all_song_ids = set(model.wv.index_to_key)
emotion_lookup = {
    row["track_id"]: row[emotion_cols].values
    for _, row in df.drop_duplicates("track_id").iterrows()
}
w2v_lookup = {
    song: model.wv[song] for song in model.wv.index_to_key
}

results = []

for pid, group in tqdm(df.groupby("pid"), desc="Evaluating playlists"):
    track_ids = group["track_id"].tolist()
    if len(track_ids) < 5:
        continue

    # Keep 75% of playlist and put 25% in held out
    sample_size = int(len(track_ids) * 0.75)
    sampled = random.sample(track_ids, sample_size)
    held_out = [song for song in track_ids if song not in sampled]

    # Get playlist vectors
    w2v_vecs = [w2v_lookup[s] for s in sampled if s in w2v_lookup]
    emo_vecs = [emotion_lookup[s] for s in sampled if s in emotion_lookup]

    if not w2v_vecs or not emo_vecs:
        continue

    playlist_vec = np.mean(w2v_vecs, axis=0)
    playlist_emo_vec = np.mean(emo_vecs, axis=0)

    combined_scores = []

    for song in all_song_ids - set(sampled):
        if song not in w2v_lookup or song not in emotion_lookup:
            continue

        word_sim = 1 - cosine(playlist_vec, w2v_lookup[song])
        emo_sim = 1 - cosine(playlist_emo_vec, emotion_lookup[song])
        combined_sim = (word_sim + emo_sim) / 2

        combined_scores.append((song, combined_sim))

    top_5 = [s for s, _ in sorted(combined_scores, key=lambda x: x[1], reverse=True)[:5]]
    hits = sum([1 for s in top_5 if s in held_out])

    results.append({
        "pid": pid,
        "n_total": len(track_ids),
        "n_sampled": len(sampled),
        "n_held_out": len(held_out),
        "hits_in_top_5": hits
    })

# --- Summary
results_df = pd.DataFrame(results)
n_with_hits = (results_df["hits_in_top_5"] >= 1).sum()
prop_with_hits = n_with_hits / len(results_df)
print(f"✅ Playlists with at least 1 correct song in top 5: {n_with_hits}/{len(results_df)} ({prop_with_hits:.2%})")
```

```python
from gensim.models import Word2Vec
from scipy.spatial.distance import cosine
from tqdm import tqdm
import numpy as np
import pandas as pd
import random

emotion_cols = ["admiration","amusement","anger","annoyance","approval","caring","confusion", "curiosity", "desire", "disappointment", "disapproval","disgust", "embarrassment", "excitement", "fear","gratitude","grief","joy", "love","nervousness","optimism","pride","realization","relief","remorse","sadness","surprise","neutral"]

# Word2Vec training
playlist_sequences = df.groupby("pid")["track_id"].apply(list).tolist()
model = Word2Vec(sentences=playlist_sequences,
                 vector_size=100, window=8, min_count=5,
                 workers=4, sg=1)

# Precompute lookups
all_song_ids = set(model.wv.index_to_key)
emotion_lookup = {
    row["track_id"]: row[emotion_cols].values
    for _, row in df.drop_duplicates("track_id").iterrows()
}
w2v_lookup = {
    song: model.wv[song] for song in model.wv.index_to_key
}

results = []
playlist_emotion_vectors = {} 

for pid, group in tqdm(df.groupby("pid"), desc="Evaluating playlists"):
    track_ids = group["track_id"].tolist()
    if len(track_ids) < 5:
        continue

    sample_size = int(len(track_ids) * 0.75)
    sampled = random.sample(track_ids, sample_size)
    held_out = [s for s in track_ids if s not in sampled]

    vecs = [model.wv[s] for s in sampled if s in model.wv]
    if not vecs:
        continue
    playlist_vec = np.mean(vecs, axis=0)

    emo_data = df[df["track_id"].isin(sampled)][emotion_cols]
    if emo_data.empty:
        continue
    playlist_emo_vec = emo_data.mean().values

    # ✅ Store vector for this playlist
    playlist_emotion_vectors[pid] = playlist_emo_vec

    word_scores = []
    emotion_scores = []
    combined_scores = []

    for song in all_song_ids - set(sampled):
        if song not in model.wv or song not in emo_lookup:
            continue

        word_sim = 1 - cosine(playlist_vec, model.wv[song])
        emo_sim = 1 - cosine(playlist_emo_vec, emotion_lookup[song])
        combined_sim = 0.5 * word_sim + 0.5 * emo_sim

        word_scores.append((song, word_sim))
        emotion_scores.append((song, emo_sim))
        combined_scores.append((song, combined_sim))

    top5_word = [s for s, _ in sorted(word_scores, key=lambda x: x[1], reverse=True)[:5]]
    top5_emo = [s for s, _ in sorted(emotion_scores, key=lambda x: x[1], reverse=True)[:5]]
    top5_comb = [s for s, _ in sorted(combined_scores, key=lambda x: x[1], reverse=True)[:5]]

    hits_word = sum([1 for s in top5_word if s in held_out])
    hits_emo = sum([1 for s in top5_emo if s in held_out])
    hits_comb = sum([1 for s in top5_comb if s in held_out])

    results.append({
        "pid": pid,
        "n_total": len(track_ids),
        "n_sampled": len(sampled),
        "n_held_out": len(held_out),
        "hits_word": hits_word,
        "hits_emo": hits_emo,
        "hits_comb": hits_comb
    })
emotion_vec_df = pd.DataFrame.from_dict(
    playlist_emotion_vectors,
    orient="index",
    columns=emotion_cols
).reset_index().rename(columns={"index": "pid"})
```

```python
import numpy as np
import pandas as pd
import random
from gensim.models import Word2Vec
from sklearn.cluster import KMeans
from scipy.spatial.distance import cosine
from tqdm import tqdm

# --------------------------
# 1. Define emotion columns
# --------------------------
emotion_cols = ['admiration', 'anger', 'annoyance', 'approval', 'curiosity', 'desire', 'disappointment', 'love', 'sadness', 'neutral']
# --------------------------
# 2. Create track_id if needed
# --------------------------
df["track_id"] = df["track_name"] + " - " + df["artist_name"]

# --------------------------
# 3. Train Word2Vec model
# --------------------------
playlist_sequences = df.groupby("pid")["track_id"].apply(list).tolist()
model = Word2Vec(sentences=playlist_sequences, vector_size=100, window=8, min_count=5, workers=4, sg=1)
all_song_ids = set(model.wv.index_to_key)

# --------------------------
# 4. Build emotion vector lookup
# --------------------------
emotion_lookup = {
    row["track_id"]: row[emotion_cols].values
    for _, row in df.drop_duplicates("track_id").iterrows()
}

# --------------------------
# 5. KMeans + Combined Scoring Loop
# --------------------------
results = []

for pid, group in tqdm(df.groupby("pid"), desc="KMeans Cluster Scoring"):
    track_ids = group["track_id"].tolist()
    if len(track_ids) < 6:
        continue

    # 75/25 split
    sample_size = int(len(track_ids) * 0.75)
    sampled = random.sample(track_ids, sample_size)
    held_out = set(track_ids) - set(sampled)

    # Cluster Word2Vec vectors
    sampled_w2v = np.array([model.wv[s] for s in sampled if s in model.wv])
    if sampled_w2v.shape[0] < 3:
        continue
    kmeans_w2v = KMeans(n_clusters=3, random_state=42, n_init=10).fit(sampled_w2v)
    centers_w2v = kmeans_w2v.cluster_centers_

    # Cluster emotion vectors
    sampled_emo = np.array([emotion_lookup[s] for s in sampled if s in emotion_lookup])
    if sampled_emo.shape[0] < 3:
        continue
    kmeans_emo = KMeans(n_clusters=3, random_state=42, n_init=10).fit(sampled_emo)
    centers_emo = kmeans_emo.cluster_centers_

    # Scoring function
    def max_sim_to_centers(vec, centers):
        return max(1 - cosine(vec, center) for center in centers)

    combined_scores = []
    for song in all_song_ids - set(sampled):
        if song in model.wv and song in emotion_lookup:
            word_sim = max_sim_to_centers(model.wv[song], centers_w2v)
            emo_sim = max_sim_to_centers(emotion_lookup[song], centers_emo)
            combined_sim = 0.5 * word_sim + 0.5 * emo_sim
            combined_scores.append((song, combined_sim))

    # Evaluate top 5
    top5 = [s for s, _ in sorted(combined_scores, key=lambda x: x[1], reverse=True)[:5]]
    hits = sum(s in held_out for s in top5)

    results.append({
        "pid": pid,
        "n_total": len(track_ids),
        "n_sampled": len(sampled),
        "n_held_out": len(held_out),
        "hits_kmeans_comb": hits
    })

# --------------------------
# 6. Save or inspect results
# --------------------------
kmeans_results_df = pd.DataFrame(results)

print(f"\n✅ Proportion of playlists with ≥1 hit in top 5: {(kmeans_results_df['hits_kmeans_comb'] >= 1).mean():.2%}")
```

```python
from gensim.models import Word2Vec
from sklearn.cluster import KMeans
from scipy.spatial.distance import cosine
from tqdm import tqdm
import numpy as np
import pandas as pd
import random

# 1. Define emotion columns
emotion_cols = [
    "admiration","amusement","anger","annoyance","approval","caring",
    "confusion","curiosity","desire","disappointment","disapproval",
    "disgust","embarrassment","excitement","fear","gratitude","grief",
    "joy","love","nervousness","optimism","pride","realization","relief",
    "remorse","sadness","surprise","neutral"
]

# 2. Make sure track_id exists
df["track_id"] = df["track_name"] + " - " + df["artist_name"]

# 3. Train Word2Vec model on playlist sequences
playlist_sequences = df.groupby("pid")["track_id"].apply(list).tolist()
model = Word2Vec(sentences=playlist_sequences, vector_size=100, window=8, min_count=5, workers=4, sg=1)
all_song_ids = set(model.wv.index_to_key)

# 4. Precompute emotion vector lookup
emotion_lookup = {
    row["track_id"]: row[emotion_cols].values
    for _, row in df.drop_duplicates("track_id").iterrows()
}

# 5. Evaluation
results = []

for pid, group in tqdm(df.groupby("pid"), desc="Evaluating playlists"):
    track_ids = group["track_id"].tolist()
    if len(track_ids) < 6:
        continue

    # 75/25 split
    sample_size = int(len(track_ids) * 0.75)
    sampled = random.sample(track_ids, sample_size)
    held_out = set(track_ids) - set(sampled)

    # Mean playlist vectors
    vecs = [model.wv[s] for s in sampled if s in model.wv]
    emo_vecs = [emotion_lookup[s] for s in sampled if s in emotion_lookup]
    if not vecs or not emo_vecs:
        continue
    playlist_vec = np.mean(vecs, axis=0)
    playlist_emo_vec = np.mean(emo_vecs, axis=0)

    # KMeans clustering
    if len(vecs) < 3 or len(emo_vecs) < 3:
        continue
    kmeans_w2v = KMeans(n_clusters=3, random_state=42, n_init=10).fit(vecs)
    kmeans_emo = KMeans(n_clusters=3, random_state=42, n_init=10).fit(emo_vecs)
    centers_w2v = kmeans_w2v.cluster_centers_
    centers_emo = kmeans_emo.cluster_centers_

    def max_sim_to_centers(vec, centers):
        return max(1 - cosine(vec, center) for center in centers)

    # Score candidates
    word_scores = []
    emo_scores = []
    comb_scores = []
    kmeans_word_scores = []
    kmeans_emo_scores = []
    kmeans_comb_scores = []

    for song in all_song_ids - set(sampled):
        if song not in model.wv or song not in emotion_lookup:
            continue

        # Vector-based
        wsim = 1 - cosine(playlist_vec, model.wv[song])
        esim = 1 - cosine(playlist_emo_vec, emotion_lookup[song])
        word_scores.append((song, wsim))
        emo_scores.append((song, esim))
        comb_scores.append((song, 0.5 * wsim + 0.5 * esim))

        # KMeans-based
        kwsim = max_sim_to_centers(model.wv[song], centers_w2v)
        kesim = max_sim_to_centers(emotion_lookup[song], centers_emo)
        kmeans_word_scores.append((song, kwsim))
        kmeans_emo_scores.append((song, kesim))
        kmeans_comb_scores.append((song, 0.5 * kwsim + 0.5 * kesim))

    # Top 5 for each model
    top5_word = [s for s, _ in sorted(word_scores, key=lambda x: x[1], reverse=True)[:5]]
    top5_emo = [s for s, _ in sorted(emo_scores, key=lambda x: x[1], reverse=True)[:5]]
    top5_comb = [s for s, _ in sorted(comb_scores, key=lambda x: x[1], reverse=True)[:5]]
    top5_kword = [s for s, _ in sorted(kmeans_word_scores, key=lambda x: x[1], reverse=True)[:5]]
    top5_kemo = [s for s, _ in sorted(kmeans_emo_scores, key=lambda x: x[1], reverse=True)[:5]]
    top5_kcomb = [s for s, _ in sorted(kmeans_comb_scores, key=lambda x: x[1], reverse=True)[:5]]

    # Count hits
    results.append({
        "pid": pid,
        "n_total": len(track_ids),
        "n_sampled": len(sampled),
        "n_held_out": len(held_out),
        "hits_word": sum(s in held_out for s in top5_word),
        "hits_emo": sum(s in held_out for s in top5_emo),
        "hits_comb": sum(s in held_out for s in top5_comb),
        "hits_kmeans_word": sum(s in held_out for s in top5_kword),
        "hits_kmeans_emo": sum(s in held_out for s in top5_kemo),
        "hits_kmeans_comb": sum(s in held_out for s in top5_kcomb)
    })

# Convert to DataFrame
results_df = pd.DataFrame(results)
```

```python
import seaborn as sns
from gensim.models import Word2Vec
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans
from scipy.spatial.distance import cosine

# Use your `df`, `model`, and `emotion_lookup` from before
sample_pids = df["pid"].unique()
selected_pids = np.random.choice(sample_pids, 4, replace=False)

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.flatten()

for i, pid in enumerate(selected_pids):
    track_ids = df[df["pid"] == pid]["track_id"].tolist()
    if len(track_ids) < 6:
        continue

    # Split
    sample_size = int(len(track_ids) * 0.75)
    sampled = random.sample(track_ids, sample_size)
    held_out = [s for s in track_ids if s not in sampled]

    # Word vectors
    sampled_vecs = [model.wv[s] for s in sampled if s in model.wv]
    held_out_vecs = [model.wv[s] for s in held_out if s in model.wv]
    if len(sampled_vecs) < 3 or not held_out_vecs:
        continue

    # --- Regular mean predictions
    playlist_vec = np.mean(sampled_vecs, axis=0)
    sim_scores = [(s, 1 - cosine(playlist_vec, model.wv[s]))
                  for s in model.wv.index_to_key if s not in sampled]
    pred_regular = [model.wv[s] for s, _ in sorted(sim_scores, key=lambda x: x[1], reverse=True)[:5]]

    # --- KMeans Word2Vec
    kmeans_word = KMeans(n_clusters=3, random_state=42).fit(sampled_vecs)
    centers_word = kmeans_word.cluster_centers_
    def max_sim(vec, centers): return max(1 - cosine(vec, c) for c in centers)
    # Compute all sims
    kmeans_word_sim_scores = [
        (s, max_sim(model.wv[s], centers_word))
        for s in model.wv.index_to_key if s not in sampled
    ]

# Sort and get top 5
    top_kmeans_word = sorted(kmeans_word_sim_scores, key=lambda x: x[1], reverse=True)[:5]
    pred_kmeans_word = [model.wv[s] for s, _ in top_kmeans_word]


    # --- KMeans Emotion
    sampled_emo = [emotion_lookup[s] for s in sampled if s in emotion_lookup]
    if len(sampled_emo) < 3:
        continue
    kmeans_emo = KMeans(n_clusters=3, random_state=42).fit(sampled_emo)
    centers_emo = kmeans_emo.cluster_centers_
    def max_sim_emo(s):
        if s not in emotion_lookup:
            return -1
        return max(1 - cosine(emotion_lookup[s], c) for c in centers_emo)
    pred_kmeans_emo = [
        model.wv[s] for s in model.wv.index_to_key
        if s not in sampled and s in emotion_lookup and max_sim_emo(s) >= 0
    ][:5]

    # Diagnostics
    print(f"PID {pid} → REG: {len(pred_regular)}, KM-W: {len(pred_kmeans_word)}, KM-E: {len(pred_kmeans_emo)}")

    # Fallback dummy point if any are empty
    dummy = np.zeros(model.vector_size)
    if not pred_regular: pred_regular = [dummy]
    if not pred_kmeans_word: pred_kmeans_word = [dummy]
    if not pred_kmeans_emo: pred_kmeans_emo = [dummy]

    # Build vector-label pairs
    vectors = (
        [(v, "held-in") for v in sampled_vecs] +
        [(v, "held-out") for v in held_out_vecs] +
        [(v, "pred-regular") for v in pred_regular] +
        [(v, "pred-kmeans-word") for v in pred_kmeans_word] +
        [(v, "pred-kmeans-emo") for v in pred_kmeans_emo]
    )
    vecs, labels = zip(*vectors)
    vecs = np.array(vecs)

    # TSNE
    tsne_2d = TSNE(n_components=2, perplexity=5, random_state=42).fit_transform(vecs)

    # Plot
    ax = axes[i]
    palette = {
        "held-in": "orange",
        "held-out": "blue",
        "pred-regular": "red",
        "pred-kmeans-word": "green",
        "pred-kmeans-emo": "purple"
    }
    for label in set(labels):
        points = tsne_2d[np.array(labels) == label]
        ax.scatter(points[:, 0], points[:, 1], label=label, color=palette[label])
    ax.set_title(f"Playlist {pid}")
    ax.legend()

plt.suptitle("TSNE of Held-In, Held-Out, and Predicted Songs (4 Playlists)", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
    
```

```python
# Re-import necessary modules after code execution environment reset
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans
from scipy.spatial.distance import cosine
import numpy as np
import pandas as pd
from tqdm import tqdm
import ace_tools as tools

# Dummy df, model, emotion_lookup definitions (must be replaced with user's actual data)
# For demonstration purposes, this will raise an error unless df, model, and emotion_lookup are defined externally
try:
    # Emotion columns
    emotion_cols = [
        "admiration", "amusement", "anger", "annoyance", "approval", "caring",
        "confusion", "curiosity", "desire", "disappointment", "disapproval",
        "disgust", "embarrassment", "excitement", "fear", "gratitude", "grief",
        "joy", "love", "nervousness", "optimism", "pride", "realization", "relief",
        "remorse", "sadness", "surprise", "neutral"
    ]

    def optimal_kmeans(vectors, max_k=6):
        """Select the best k using silhouette score."""
        if len(vectors) < 3:
            return KMeans(n_clusters=1, random_state=42, n_init=10).fit(vectors)
        best_k = 1
        best_score = -1
        best_model = None
        for k in range(2, min(max_k + 1, len(vectors))):
            model_k = KMeans(n_clusters=k, random_state=42, n_init=10)
            labels = model_k.fit_predict(vectors)
            score = silhouette_score(vectors, labels)
            if score > best_score:
                best_k = k
                best_score = score
                best_model = model_k
        if best_model is None:
            best_model = KMeans(n_clusters=1, random_state=42, n_init=10).fit(vectors)
        return best_model

    def max_sim_to_centers(vec, centers):
        return max(1 - cosine(vec, center) for center in centers)

    all_song_ids = set(model.wv.index_to_key)
    results = []

    for pid, group in tqdm(df.groupby("pid"), desc="Evaluating playlists with adaptive KMeans"):
        track_ids = group["track_id"].tolist()
        if len(track_ids) < 6:
            continue

        sample_size = int(len(track_ids) * 0.75)
        sampled = np.random.choice(track_ids, sample_size, replace=False).tolist()
        held_out = list(set(track_ids) - set(sampled))

        # --- Word2Vec clustering
        sampled_vecs = [model.wv[s] for s in sampled if s in model.wv]
        if len(sampled_vecs) < 3:
            continue
        kmeans_word = optimal_kmeans(sampled_vecs)
        centers_word = kmeans_word.cluster_centers_

        # --- Emotion clustering
        sampled_emo = [emotion_lookup[s] for s in sampled if s in emotion_lookup]
        if len(sampled_emo) < 3:
            continue
        kmeans_emo = optimal_kmeans(sampled_emo)
        centers_emo = kmeans_emo.cluster_centers_

        # --- Score candidates
        pred_word_scores = [
            (s, max_sim_to_centers(model.wv[s], centers_word))
            for s in all_song_ids - set(sampled)
            if s in model.wv
        ]
        pred_emo_scores = [
            (s, max_sim_to_centers(emotion_lookup[s], centers_emo))
            for s in all_song_ids - set(sampled)
            if s in emotion_lookup
        ]
        pred_comb_scores = {
            s: 0.5 * dict(pred_word_scores).get(s, 0) + 0.5 * dict(pred_emo_scores).get(s, 0)
            for s in all_song_ids - set(sampled)
            if s in model.wv and s in emotion_lookup
        }

        top5_word = [s for s, _ in sorted(pred_word_scores, key=lambda x: x[1], reverse=True)[:5]]
        top5_emo = [s for s, _ in sorted(pred_emo_scores, key=lambda x: x[1], reverse=True)[:5]]
        top5_comb = [s for s, _ in sorted(pred_comb_scores.items(), key=lambda x: x[1], reverse=True)[:5]]

        results.append({
            "pid": pid,
            "n_total": len(track_ids),
            "n_sampled": len(sampled),
            "n_held_out": len(held_out),
            "hits_kmeans_word": sum([s in held_out for s in top5_word]),
            "hits_kmeans_emo": sum([s in held_out for s in top5_emo]),
            "hits_kmeans_comb": sum([s in held_out for s in top5_comb]),
        })

    results_df_adaptive = pd.DataFrame(results)
    tools.display_dataframe_to_user(name="Adaptive KMeans Playlist Evaluation", dataframe=results_df_adaptive)

except Exception as e:
    print("Please make sure df, model, and emotion_lookup are defined in your environment.")
    print(str(e))
```
